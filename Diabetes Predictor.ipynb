```python
import pandas as pd
import joblib

try:
    model, scaler, saved_cols = joblib.load("diabetes_gb.pkl")
    scaler = scaler.set_output(transform='pandas')

except FileNotFoundError:
    print("Error: 'diabetes_gb.pkl' not found. Please ensure the model file is in the same directory.")
    raise

def predict_diabetes():
    """Collects user input via the console, makes a prediction, and prints the result."""
    print("\n" + "="*50)
    print("Diabetes Prediction")
    print("Please enter the patient's data:")
    print("="*50)

    try:
        pregnancies = int(input("Pregnancies (0-20, e.g., 2): "))
        glucose = int(input("Glucose (0-200, e.g., 100): "))
        blood_pressure = int(input("Blood Pressure (0-150, e.g., 70): "))
        skin_thickness = int(input("Skin Thickness (0-100, e.g., 20): "))
        insulin = int(input("Insulin (0-900, e.g., 80): "))
        bmi = float(input("BMI (0.0-70.0, e.g., 25.0): "))
        dpf = float(input("Diabetes Pedigree Function (0.0-3.0, e.g., 0.5): "))
        age = int(input("Age (1-120, e.g., 30): "))
    except ValueError:
        print("\n--- ERROR ---")
        print("Invalid input. All values must be numeric. Please try again.")
        return

    input_data = pd.DataFrame(
        [[pregnancies, glucose, blood_pressure, skin_thickness, insulin, bmi, dpf, age]],
        columns=saved_cols
    )

    input_scaled = scaler.transform(input_data)

    prediction = model.predict(input_scaled)[0]

    print("\n" + "="*20)
    print("PREDICTION RESULT")
    print("="*20)
    if prediction == 1:
        print("The patient is likely to have Diabetes.")
    else:
        print("The patient is unlikely to have Diabetes.")
    print("="*50 + "\n")

while True:
    predict_diabetes()

    repeat = input("Do you want to predict again? (yes/no or y/n): ")
    if repeat.strip().lower().startswith('y'):
        continue
    else:
        print("Thank you for using the Diabetes Prediction tool. Goodbye!")
        break
```
